```python
df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob'],
    'Age': [25, 30],
    'City': ['New York', 'London'],
    'salary': [70000, 80000]
})
df1
```

```python
df2 = pd.DataFrame({
    'Name': ['Charlie', 'David', 'Grace', 'Heidi'],
    'Age': [35, 40, 29, 31],
    'City': ['Paris', 'Tokyo', 'Rome', 'Madrid'],
    'salary': [90000, 100000, 85000, 95000]
})
df2
```

```python
df = pd.concat([df1, df2])
df
```

```python
df['Rank'] = df['salary'].rank(ascending=False, method='dense')
df
```

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Math': [85, 90, 78],
    'Science': [92, 88, 95],
    'English': [80, 85, 82]
})

df
```

```python

melted_df = pd.melt(df, id_vars = ['Name'], var_name = 'Subject', value_name = 'Score')
melted_df
```

```python
pivot_table_df = pd.pivot_table(melted_df, index='Name', values='Score', aggfunc='sum')
pivot_table_df
```
