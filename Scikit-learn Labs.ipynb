## Loading and fetching Data


```python
import numpy as np
from sklearn.datasets import load_breast_cancer
```

   * 0: Malignant (Cancerous)
   * 1: Benign (Non-cancerous)

```python
data = load_breast_cancer()
X, y = load_breast_cancer(return_X_y=True)
print(X.shape, y.shape)
```

```python
data_df = load_breast_cancer(as_frame=True)
df = data_df.frame

print(df.shape)
df
```

## Data Splitting

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=44)
print(X_train.shape, y_train.shape)
```

```python
print(y_test, y_train)
```

```python
plt.bar(["y_train", "y_test"], [np.bincount(y_train)[0], np.bincount(y_test)[0]], label="0")
plt.bar(["y_train", "y_test"], [np.bincount(y_train)[1], np.bincount(y_test)[1]], label="1", bottom=[np.bincount(y_train)[0], np.bincount(y_test)[0]])
plt.legend()
plt.title("Showing the ratio of y_train and y_test")
plt.tight_layout()
plt.show()
```

## Data Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(X_train_scaled)


print(X_test_scaled)
```

## Training

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=10000, random_state=44)
model.fit(X_train_scaled, y_train)
y_pred_logreg = model.predict(X_test_scaled)
print(model.score(X_test_scaled, y_test))
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis()

lda.fit(X_train_scaled, y_train)
y_pred_lda = lda.predict(X_test_scaled)
print(f"Accuracy of LDA: {lda.score(X_test_scaled, y_test)}")
```

## PCA Visulization

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_train_scaled)
print(X_pca.shape)
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_train, cmap='viridis', edgecolor='k', alpha=0.7)
plt.title("PCA Projection of Breast Cancer Dataset")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.colorbar(label='Class Label')
plt.tight_layout()
plt.show()
```

```python
print(f"Variance explained by each component: {pca.explained_variance_ratio_}")

total_variance_explained = np.sum(pca.explained_variance_ratio_)  *100
print(f"Total variance explained by the 2 components: {total_variance_explained:.2f}%")
```

## Advanced Models: SVM and Random Forest
To handle non-linear relationships and edge cases, we'll try a Support Vector Machine (with RBF kernel) and a Random Forest Classifier.

```python
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

# 1. Support Vector Machine (RBF Kernel)
# probability=True allows us to use predict_proba later for ROC curves
svm_model = SVC(kernel='rbf', probability=True, random_state=44)
svm_model.fit(X_train_scaled, y_train)
y_pred_svm = svm_model.predict(X_test_scaled)

# 2. Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=44)
rf_model.fit(X_train_scaled, y_train)
y_pred_rf = rf_model.predict(X_test_scaled)

print(f"SVM Accuracy: {svm_model.score(X_test_scaled, y_test):.4f}")
print(f"Random Forest Accuracy: {rf_model.score(X_test_scaled, y_test):.4f}")
```

## Decision Boundary Visualization

```python
# 1. Create a meshgrid in the 2D PCA space
x_min, x_max = X_pca[:, 0].min() - 1, X_pca[:, 0].max() + 1
y_min, y_max = X_pca[:, 1].min() - 1, X_pca[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1), np.arange(y_min, y_max, 0.1))

# 2. Map the 2D PCA grid points back to the 30D feature space
grid_points_pca = np.c_[xx.ravel(), yy.ravel()]
grid_points_30d = pca.inverse_transform(grid_points_pca)

# 3. Use the original 30-feature model to predict
Z = model.predict(grid_points_30d)
Z = Z.reshape(xx.shape)

# 4. Plot the results
plt.figure(figsize=(10, 7))
plt.contourf(xx, yy, Z, alpha=0.3, cmap='viridis')
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_train, edgecolors='k', cmap='viridis', alpha=0.8)

plt.title("30-Feature Decision Boundary projected onto PCA Space")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```

### Side-by-Side Model Comparison
We can now compare how the three different models (Linear vs. Curved vs. Boxy) handle the exact same data.

```python
# 1. Setup the grid (reusing the logic from the previous cell)
x_min, x_max = X_pca[:, 0].min() - 1, X_pca[:, 0].max() + 1
y_min, y_max = X_pca[:, 1].min() - 1, X_pca[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1), np.arange(y_min, y_max, 0.1))

# 2. Map the 2D grid points back to 30D
grid_points_pca = np.c_[xx.ravel(), yy.ravel()]
grid_points_30d = pca.inverse_transform(grid_points_pca)

# 3. Create a side-by-side visualization
fig, axes = plt.subplots(1, 3, figsize=(20, 6))
models = [model, svm_model, rf_model]
titles = ["Logistic Regression (Linear)", "SVM (RBF Kernel)", "Random Forest (Tree-based)"]

for ax, clf, title in zip(axes, models, titles):
    # Predict
    Z = clf.predict(grid_points_30d)
    Z = Z.reshape(xx.shape)
    
    # Plot background
    ax.contourf(xx, yy, Z, alpha=0.3, cmap='viridis')
    # Plot points
    ax.scatter(X_pca[:, 0], X_pca[:, 1], c=y_train, edgecolors='k', cmap='viridis', s=20)
    
    ax.set_title(title)
    ax.set_xlabel("PC1")
    ax.set_ylabel("PC2")

plt.tight_layout()
plt.show()
```

## Evaluation

```python
from sklearn.metrics import confusion_matrix, roc_curve, roc_auc_score, classification_report
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Confusion Matrix Heatmap
cm = confusion_matrix(y_test, y_pred_logreg)
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Malignant', 'Benign'], 
            yticklabels=['Malignant', 'Benign'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix: Logistic Regression')
plt.show()

# 2. Classification Report
print("Classification Report:")
print(classification_report(y_test, y_pred_logreg, target_names=['Malignant', 'Benign']))

# 3. ROC Curve and AUC Score (Robustness)
y_prob = model.predict_proba(X_test_scaled)[:, 1]
auc = roc_auc_score(y_test, y_prob)
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f'ROC Curve (AUC = {auc:.4f})')
plt.plot([0, 1], [0, 1], 'k--') # Baseline random model
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve: Logistic Regression')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Moon data

```python
from sklearn.datasets import make_moons
from matplotlib import pyplot as plt

data = make_moons(n_samples=1000, noise=0.1, random_state=44)

plt.scatter(data[0][:, 0], data[0][:, 1], c=data[1], edgecolors='k', cmap='viridis', s=20)
plt.title("Showing Moon Data")
plt.tight_layout()
plt.show()
```

### Data Splitting

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(data[0], data[1], test_size=0.2, random_state=44)
```

### Preprocessing

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### Training

#### Sperating Data Using LDA

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis()

lda.fit(X_train_scaled, y_train)

y_pred_lda = lda.predict(X_test_scaled)
print(f"Accuracy of LDA: {lda.score(X_test_scaled, y_test)}")
```

```python
from sklearn.inspection import DecisionBoundaryDisplay

fig, ax = plt.subplots(figsize=(8, 6))
display_moon = DecisionBoundaryDisplay.from_estimator(lda, X_train_scaled, ax=ax, response_method='predict', cmap='viridis', alpha=0.3)
ax.scatter(X_train_scaled[:, 0], X_train_scaled[:, 1], c=y_train, edgecolors='k', cmap='viridis', s=20)
ax.set_title("LDA Decision Boundary on Moon Data")
plt.tight_layout()
plt.show()
```

#### Sperating Data using SVC (Support Vector Classifier)

```python
from sklearn.svm import SVC

svm = SVC(kernel='rbf', random_state=44)

svm.fit(X_train_scaled, y_train)

svm_pred = svm.predict(X_test_scaled)
print(f"SVM Accuracy on Moon Data: {svm.score(X_test_scaled, y_test):.4f}")
```

```python
from sklearn.inspection import DecisionBoundaryDisplay

fig, ax = plt.subplots(figsize=(8, 6))
display_moon_svm = DecisionBoundaryDisplay.from_estimator(svm, X_train_scaled, ax=ax, response_method='predict', cmap='viridis', alpha=0.3)
ax.scatter(X_train_scaled[:, 0], X_train_scaled[:, 1], c=y_train, edgecolors='k', cmap='viridis', s=20)
ax.set_title("SVM Decision Boundary on Moon Data")
plt.tight_layout()
plt.show()
```
