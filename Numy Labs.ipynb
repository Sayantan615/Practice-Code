```python
# How to create a n*m matrix of 1 to n*m
n, m = 5, 12
a = np.arange(1, n * m + 1).reshape(n, m)
print(a)
```

```python
print(a[1, 5])
print(a[1, :])
print(a[0, 1:12:2])

mask = a < 10
filtered_a = a[mask]
print(filtered_a)
```

```python
print(np.any(a < 29, axis=1))
```

```python
rows = [1, 2]
cols = [5, 6]
print(a[rows, cols])
```

```python
np.where(a < 29, 'Yes', 'No')
```
