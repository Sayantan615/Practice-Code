### Matrix Multiplication

- The inner dimensions must match
- The resulting matrix has the shape of the outer dimensions

```python
t1 = tf.constant([[1,2,5],[7,2,1],[3,3,3]])
t2 = tf.constant([[3,5],[6,7],[1,8]])

# create 3D tensors t3 of shape (2, 3, 4), t4 of shape (4, 3, 2)
t3 = tf.random.normal(shape=(8, 3, 4))
t4 = tf.random.normal(shape=(8, 4, 2))

tf.linalg.matmul(t1, t2)  # you can drop .linalg
t3.shape, t4.shape
tf.matmul(t3, t4)
```

```python
# Two same shaped metrices
X = tf.constant([[1,2],[3,4],[5,6]])
Y = tf.constant([[7,8],[9,10],[11,12]])
print(X.shape, Y.shape)

# Reshaping y
tf.reshape(Y, shape=(2,3))

# Then multiplying
print(tf.matmul(X, tf.reshape(Y, shape=(2,3))))
print(tf.matmul(tf.reshape(X, shape=(2,3)), Y))
```

```python
## We can achieve the same using tf.transpose

print(tf.matmul(tf.transpose(X), Y))
```

### The dot product
Matrix multiplication is also referred to as the dot product

this can be peformed using matrix multiplications using

- tf.matmul()
- tf.tensordot() - what is tensort dot does

generally when performing matrix multiplication on two tesors and one of the axes doesn't line up we will transpose rather than reshape to get satify dot product conditions

```python
# Usage of tensordot
tf.tensordot(tf.transpose(X), Y, axes=1)
```

```python
E = tf.constant(np.random.randint(0, 100, size=50))
E
```

### tensorflow Probablity
