- What is difference between tf.constant and tf.variable
- when in doubt, use tf.constant and change it later if needed, how

```python
# tensor variables
changeable_tensor = tf.Variable([10,7])
unchangeable_tensor = tf.constant([10,7])
changeable_tensor,unchangeable_tensor
```

```python
changeable_tensor[0] = 11
changeable_tensor
```

```python
changeable_tensor[0].assign(8)
changeable_tensor
```

```python
unchangeable_tensor[0].assign(8)
unchangeable_tensor
```
